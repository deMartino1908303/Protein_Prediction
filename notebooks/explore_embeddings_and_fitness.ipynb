```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = np.load("embeddings.pkl", allow_pickle=True)
# if needed: data = data.item()

vals = np.asarray(data.fitness).astype(float)

# compute CDF
sorted_vals = np.sort(vals)
cdf = np.arange(1, len(sorted_vals)+1) / len(sorted_vals)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(7,5))
plt.plot(sorted_vals, cdf, linewidth=2)

plt.xlabel("Fitness value")
plt.ylabel("CDF")
plt.title("CDF of Fitness Values")

plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(7,5))
sns.histplot(
    sorted_vals,
    bins=40,
    kde=True,
    stat="density",
    edgecolor="black",
    linewidth=0.5
)

plt.xlabel("Fitness value")
plt.ylabel("Density")
plt.title("Distribution of Fitness Values")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# load data
data = np.load("embeddings.pkl", allow_pickle=True)
# if needed: data = data.item()
data["emb_dim"] = pd.Series(asdnjas)
df = pd.DataFrame({
    "emb_dim": data.emb_dim,
    "fitness": data.fitness
})

dims = [256, 512, 768]

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8,6))

colors = {
    256: "tab:blue",
    512: "tab:orange",
    768: "tab:green"
}

for d in dims:
    vals = df.loc[df["emb_dim"] == d, "fitness"].astype(float).to_numpy()

    sns.histplot(
        vals,
        bins=40,
        kde=True,
        stat="density",
        edgecolor="black",
        linewidth=0.5,
        color=colors[d],
        alpha=0.35,
        label=f"emb_dim = {d}"
    )

plt.xlabel("Fitness value")
plt.ylabel("Density")
plt.title("Distribution of Fitness Values by emb_dim")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# 1. Fix SettingWithCopyWarning: Ensure df_clean is a completely independent copy
df_clean = df_clean.copy()

# --- BINS INFO ---
interval = [-np.inf, -2, -.5, .5, 2, np.inf]
inter_names = ["Destroied", "Deleterious", "Neutral", "Beneficial", "Improved"]

# Assign the new category
df_clean["fitness_category"] = pd.cut(
    df_clean["fitness"], 
    bins=interval, 
    labels=inter_names
)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

# 2. Fix FutureWarning: Assign 'hue' to the same variable as 'x' and disable legend
sns.boxplot(
    data=df_clean,
    x="fitness_category",
    y="fitness",
    hue="fitness_category",  # Explicitly map color to category
    palette="vlag",
    legend=False             # Hide the redundant legend
)

plt.xlabel("Fitness Category")
plt.ylabel("Fitness Value")
plt.title("Fitness Distribution by Category")

plt.tight_layout()
plt.show()
```

```python
df_plot = df_clean.copy()

interval = [-np.inf, -2, -.5, .5, 2, np.inf]
inter_names = ["Destroied", "Deleterious", "Neutral", "Beneficial", "Improved"]
custom_colors = ["#320404", "#B01818", "#AFAFAF", "#025C11", "#00DB2F"]

# Create the category column needed for 'hue'
df_plot["fitness_category"] = pd.cut(
    df_plot["fitness"], 
    bins=interval, 
    labels=inter_names
)

# 2. Create the Overlapping Plot
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

sns.histplot(
    data=df_plot,
    x="fitness",
    hue="fitness_category",  # Splits the data into the 5 groups
    palette=custom_colors,   # Applies your specific colors
    bins=50,                 # Increase bins slightly for better resolution across groups
    multiple="layer",        # THE KEY PARAMETER: layers them transparently
    edgecolor="black",       # Adds definition to the bars
    linewidth=0.3,
    alpha=0.6                # Makes bars semi-transparent to see overlaps
)

# Adjust labels and title
plt.xlabel("Fitness value")
plt.ylabel("Count") # Changed to Count as it's clearer for overlapping histograms
plt.title("Overlapping Distribution of Fitness Categories")

# Move legend slightly outside if it covers data, or keep it inside
sns.move_legend(plt.gca(), "upper right", title="Category")

plt.tight_layout()
plt.show()
```

```python
# --- BINS INFO ---
interval = [-np.inf, -2, -.5, .5, 2, np.inf]
inter_names = ["Destroied", "Deleterious", "Neutral", "Beneficial", "Improved"]
custom_colors = ["#320404", "#B01818", "#AFAFAF", "#025C11", "#00DB2F"]
#  ["#5A0F0F","#D24040", "#D1EED1", "#5CEB15", "#057B1D"]
# Create the category column
df_clean["fitness_category"] = pd.cut(
    df_clean["fitness"], 
    bins=interval, 
    labels=inter_names
)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

# 2. Create the Count Plot
# sns.countplot automatically calculates the number of occurrences
ax = sns.countplot(
    data=df_clean,
    x="fitness_category",
    palette=custom_colors,
    hue="fitness_category", # Matches colors to categories
    dodge=False,            # Prevents bars from shrinking if hue is used
    order=inter_names       # Ensures categories are in the correct order
)

# 3. Clean up and Label
# Remove the legend since it duplicates the x-axis info
if ax.legend_:
    ax.legend_.remove()

# Optional: Add the exact count numbers on top of each bar
for container in ax.containers:
    ax.bar_label(container, fontsize=10)

plt.xlabel("Fitness Category")
plt.ylabel("Count of Instances")
plt.title("Number of Samples per Fitness Category")

plt.tight_layout()
plt.show()
```
